Q1 = Charger et lire le base de donnees

```python
columns = [
'CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE',
'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'MEDV'
]
df = pd.read_csv('/content/HOUSING.CSV', delimiter=r'\s+', names=columns)
```

Q3=Afficher la boîte à moustache de chaque colonne  

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(15,8))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.title("boîte à moustache ")
plt.show()
```

Q5= variables explicatives


```python
x = df.iloc[:, :-1]
y = df.iloc[:, -1]
print(x)

```

```python
sns.scatterplot(x=df['TAX'], y=df['MEDV'],color='r')
plt.title("Relation entre TAX et MEDV")
plt.show()
```

Q6= separer 70% Apprentissage et 30% test

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=42)
```

Q7=shape de cette nouvelle repartition

```python
print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
print("y_train :", y_train.shape)
print("y_test  :", y_test.shape)
```

Q8= un modèle de régression linéaire multiple sur les données d’apprentissage.

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

Q9=coefficients estimés pour le problème de régression linéaire multiple.

```python
print("Coefficients :", model.coef_)
```

Q10=modèle pour prédire les résultats sur les données de test

```python
y_pred = model.predict(X_test)
print(y_pred[:5])
```

Q11=valeur de l’erreur moyenne quadratique et celle du coefficient de détermination R²

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print("MSE =", mse)
print("R²  =", r2)
```
